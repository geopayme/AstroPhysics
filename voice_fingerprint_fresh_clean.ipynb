```python
import librosa
import numpy as np
import matplotlib.pyplot as plt
from pydub import AudioSegment
import eyed3
from google.colab import files
from pathlib import Path
```

```python
# Upload a WAV file from local system
uploaded = files.upload()
wav_file = list(uploaded.keys())[0]

# Validate file extension
if not wav_file.lower().endswith(".wav"):
    raise ValueError(f"Expected a WAV file, got: {wav_file}")
```

```python
import matplotlib.pyplot as plt
import librosa
import numpy as np

def extract_voice_fingerprint(wav_file):
    try:
        y, sr = librosa.load(wav_file, sr=None)
    except Exception as e:
        raise ValueError(f"Failed to load audio: {e}")

    mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
    fingerprint = np.mean(mfccs, axis=1)

    # Plot the fingerprint
    plt.figure(figsize=(10, 4))
    plt.plot(fingerprint, marker='o')
    plt.title("Voice Fingerprint (Mean MFCCs)")
    plt.xlabel("MFCC Coefficient Index")
    plt.ylabel("Mean Value")
    plt.grid(True)
    plt.show()

    return fingerprint
```

```python
def embed_fingerprint_in_metadata(mp3_file, fingerprint):
    fingerprint_str = ','.join([str(f) for f in fingerprint])
    audiofile = eyed3.load(mp3_file)
    if audiofile.tag is None:
        audiofile.initTag()
    audiofile.tag.comments.set(fingerprint_str)
    audiofile.tag.save(mp3_file)
```

```python
def save_new_audio_with_fingerprint(wav_file, output_file):
    audio = AudioSegment.from_wav(wav_file)
    audio.export(output_file, format='wav')
```

```python
def convert_wav_to_mp3(wav_file, mp3_file):
    audio = AudioSegment.from_wav(wav_file)
    audio.export(mp3_file, format='mp3')
```

```python
# Extract fingerprint
fingerprint = extract_voice_fingerprint(wav_file)
print("Extracted Fingerprint:", fingerprint)

# Save as new WAV
output_wav = 'output_with_fingerprint.wav'
save_new_audio_with_fingerprint(wav_file, output_wav)

# Convert to MP3 and embed fingerprint
mp3_file = 'converted_audio.mp3'
convert_wav_to_mp3(wav_file, mp3_file)
embed_fingerprint_in_metadata(mp3_file, fingerprint)
print(f"Fingerprint embedded in MP3 metadata: {mp3_file}")

# Offer downloads
files.download(output_wav)
files.download(mp3_file)
```
