```python
from sympy import symbols, Function, Derivative, simplify, expand
from IPython.display import display

# Define spacetime coordinates explicitly
x0, x1, x2, x3 = symbols('x0 x1 x2 x3')
x = (x0, x1, x2, x3)

# Define the ∞-tensor field component as a function of one coordinate (for example purposes)
psi = Function('psi')
g = Function('g')
V = Function('V')

# Pick mu = 0, nu = 1 for illustrative kinetic term
mu = 0
nu = 1

# Build kinetic term
d_psi_mu = Derivative(psi(x[mu]), x[nu])  # ∂_ν ψ(x_μ)
d_psi_nu = Derivative(psi(x[nu]), x[mu])  # ∂_μ ψ(x_ν)

L_kin = 0.5 * g(mu, nu) * d_psi_mu * d_psi_nu
L_total = L_kin - V(psi(x0))  # Potential term

# Output
print("📘 Lagrangian Density 𝓛_∞ (Evaluated at μ=0, ν=1):")
display(L_total)
```
