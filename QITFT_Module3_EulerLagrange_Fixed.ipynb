```python

# 📘 QITFT — Module 3: Euler–Lagrange Derivation (Fixed ψ(x0, x1, x2, x3))

!pip install sympy --quiet

from sympy import symbols, Function, Derivative, simplify, Add, Eq
from IPython.display import display

# Define spacetime coordinates
x0, x1, x2, x3 = symbols('x0 x1 x2 x3')
coords = (x0, x1, x2, x3)

# Define ψ as a function of all coordinates
psi = Function('psi')(*coords)

# Metric and potential
g = Function('g')
V = Function('V')

# Build full kinetic term
kinetic_terms = []

for mu in range(4):
    for nu in range(4):
        dpsi_mu = Derivative(psi, coords[nu])
        dpsi_nu = Derivative(psi, coords[mu])
        term = g(mu, nu) * dpsi_mu * dpsi_nu
        kinetic_terms.append(term)

L_kin = 0.5 * Add(*kinetic_terms)
L_pot = V(psi)
L_total = simplify(L_kin - L_pot)

print("📘 Full Lagrangian Density 𝓛_∞:")
display(L_total)

# ---------------------------------------
# 🎯 Euler–Lagrange Equation Derivation
# ---------------------------------------

# Step 1: dL/dψ
dL_dpsi = simplify(L_total.diff(psi))

# Step 2: dL/d(∂μψ) terms
dL_d_dpsi = [simplify(L_total.diff(Derivative(psi, coords[mu]))) for mu in range(4)]

# Step 3: ∂μ (dL/d(∂μψ))
euler_terms = [Derivative(dL_d_dpsi[mu], coords[mu]) for mu in range(4)]

# Final Euler–Lagrange Equation
EL_eq = simplify(Add(*euler_terms) - dL_dpsi)

print("\n📘 Euler–Lagrange Equation for ψ(x):")
display(Eq(EL_eq, 0))

# ---------------------------------------
# 🧪 Optional: Flat Metric Simplification
# ---------------------------------------

flat_metric = lambda mu, nu: 1 if mu == nu == 0 else (-1 if mu == nu else 0)
flat_terms = []

for mu in range(4):
    for nu in range(4):
        g_flat = flat_metric(mu, nu)
        dpsi_mu = Derivative(psi, coords[nu])
        dpsi_nu = Derivative(psi, coords[mu])
        term = g_flat * dpsi_mu * dpsi_nu
        flat_terms.append(term)

L_flat = simplify(0.5 * Add(*flat_terms) - V(psi))

print("\n📘 Flat Spacetime Lagrangian:")
display(L_flat)
```
