```python
from sympy import symbols, Function, Derivative, simplify, expand, Add
from IPython.display import display

# Define coordinates
x0, x1, x2, x3 = symbols('x0 x1 x2 x3')
x = (x0, x1, x2, x3)

# Field and metric function
psi = Function('psi')
g = Function('g')
V = Function('V')

# Sum over μ, ν = 0,1,2,3
kinetic_terms = []

for mu in range(4):
    for nu in range(4):
        dpsi_mu = Derivative(psi(x[mu]), x[nu])  # ∂_ν ψ(x_μ)
        dpsi_nu = Derivative(psi(x[nu]), x[mu])  # ∂_μ ψ(x_ν)
        term = g(mu, nu) * dpsi_mu * dpsi_nu
        kinetic_terms.append(term)

# Total kinetic term
L_kin = 0.5 * Add(*kinetic_terms)

# Potential
L_pot = V(psi(x0))

# Full Lagrangian
L_total = simplify(L_kin - L_pot)

# Output
print("📘 Full Lagrangian Density 𝓛_∞:")
display(L_total)
```
