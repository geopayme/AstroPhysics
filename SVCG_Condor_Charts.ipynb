# SGCV Time-of-Flight Anomaly Visualization
This notebook loads SGCV simulation data and generates:
- A histogram of time deviation (Δt)
- A scatter plot of spatial offset (Δx) vs. time deviation (Δt)
- Comparative histograms against ALP and kinetic mixing models

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the SGCV data
df = pd.read_csv('sgcv_tof_simulated_data.csv')
```

## SGCV Δt Anomaly Histogram

```python
plt.figure(figsize=(8, 6))
plt.hist(df['delta_t_s'], bins=50, color='steelblue', edgecolor='black')
plt.title('SGCV Δt Anomalies')
plt.xlabel('Time Deviation [ps]')
plt.ylabel('Counts')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## SGCV Δx vs Δt Scatter Plot

```python
plt.figure(figsize=(8, 6))
plt.scatter(df['delta_t_s'], df['delta_x_m'], alpha=0.6, s=20, color='darkgreen')
plt.title('SGCV Spatial Displacement vs Time Deviation')
plt.xlabel('Time Deviation Δt [ps]')
plt.ylabel('Spatial Displacement Δx [mm]')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Comparative Models: ALP and Kinetic Mixing

```python
# Simulate ALP and kinetic mixing models
np.random.seed(42)
alp_dt = np.random.normal(loc=0, scale=30, size=500)
kmix_dt = np.random.normal(loc=0, scale=50, size=500)

# Plot comparison
plt.figure(figsize=(8, 6))
plt.hist(df['delta_t_s'], bins=50, alpha=0.6, label='SGCV', color='blue')
plt.hist(alp_dt, bins=50, alpha=0.6, label='ALP Model', color='orange')
plt.hist(kmix_dt, bins=50, alpha=0.6, label='Kinetic Mixing', color='green')
plt.title('Comparison of Δt Distributions')
plt.xlabel('Time Deviation [ps]')
plt.ylabel('Counts')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
