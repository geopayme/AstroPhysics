```python

# 📘 QITFT — Module 7 Extension: Ricci Scalar and Einstein Equations (Fixed IndexedBase)

!pip install sympy matplotlib numpy --quiet

from sympy import symbols, Function, Eq, simplify, IndexedBase, Idx, Sum
from sympy.tensor.tensor import TensorIndexType, tensor_indices
from IPython.display import display

# ------------------------------------------------------------
# 🔷 Spacetime and Tensor Setup
# ------------------------------------------------------------

# Define spacetime dimension and indices
D = 2  # 1+1D spacetime
Lorentz = TensorIndexType('Lorentz', dummy_name='L', metric_symmetry=(1, 1))
mu, nu = tensor_indices('mu nu', Lorentz)

# IndexedBase tensors (array-style)
g = IndexedBase('g')         # Metric tensor g_{μν}
g_inv = IndexedBase('g_inv') # Inverse metric g^{μν}
R = IndexedBase('R')         # Ricci tensor R_{μν}
G = IndexedBase('G')         # Einstein tensor G_{μν}

# ------------------------------------------------------------
# 🔶 Symbolic Ricci Scalar: R = g^{μν} R_{μν}
# ------------------------------------------------------------

i = Idx('i', D)
j = Idx('j', D)

# Ricci scalar summation
R_scalar = Sum(g_inv[i, j] * R[i, j], (i, 0, D-1), (j, 0, D-1)).doit()

print("📘 Symbolic Ricci Scalar:")
display(Eq(symbols('R_scalar'), R_scalar))

# ------------------------------------------------------------
# 🔷 Einstein Field Equation: G_{μν} = R_{μν} - 1/2 g_{μν} R
# ------------------------------------------------------------

Einstein_eq = Eq(G[mu, nu], R[mu, nu] - 0.5 * g[mu, nu] * R_scalar)

print("\n📘 Einstein Field Equation:")
display(Einstein_eq)
```
