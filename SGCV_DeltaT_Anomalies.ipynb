```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("sgcv_tof_simulated_data.csv")

# Plot histogram of Δt anomalies
plt.hist(df["delta_t_s"] * 1e12, bins=50)  # Convert seconds to picoseconds
plt.title("SGCV Δt Anomalies")
plt.xlabel("Time Deviation [ps]")
plt.ylabel("Counts")
plt.grid(True)
plt.show()

```
