# SmartFlow Reactor™ PFAS Dashboard (Jupyter Version)

```python
import pandas as pd
import plotly.graph_objs as go
from IPython.display import display, Markdown
```

```python
from google.colab import files
uploaded = files.upload()
df = pd.read_csv(list(uploaded.keys())[0])
```

```python

def plot_dashboard(threshold=70):
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=df['Time_minutes'], y=df['Stage1_Mechanical_ngL'],
                             mode='lines', name='Stage 1: Mechanical', line=dict(color='gray')))
    fig.add_trace(go.Scatter(x=df['Time_minutes'], y=df['Stage2_Oxidation_ngL'],
                             mode='lines', name='Stage 2: Oxidation', line=dict(color='orange')))
    fig.add_trace(go.Scatter(x=df['Time_minutes'], y=df['Stage3_Biofiltration_ngL'],
                             mode='lines', name='Stage 3: Biofiltration', line=dict(color='green')))
    fig.add_hline(y=threshold, line=dict(color='red', dash='dash'))
    fig.update_layout(title='PFAS Destruction Kinetics – SmartFlow Reactor™',
                      xaxis_title='Time (minutes)', yaxis_title='PFAS Concentration (ng/L)', height=500)

    final_value = df['Stage3_Biofiltration_ngL'].iloc[-1]
    msg = f"⚠️ PFAS exceeds {threshold} ng/L!" if final_value > threshold else f"✅ PFAS within safe limits ({final_value:.2f} ng/L)"
    display(Markdown(f"### {msg}"))

    fig.show()

plot_dashboard()
```
