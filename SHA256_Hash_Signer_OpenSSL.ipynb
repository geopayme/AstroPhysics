# 🔐 PDF Hash Signing Tool with OpenSSL (Secure GPG-like Signature)
This notebook lets you:
- Upload a `.sha256` hash file (from a signed PDF)
- Upload your **RSA private key** (`.pem` format)
- Sign the hash using `openssl dgst`
- Output a `.sig` signature file that can be verified with the public key

````python
# 📦 Install OpenSSL if needed (usually pre-installed in Colab)
!apt-get install openssl
````

````python
# 📤 Upload .sha256 file and private RSA key
from google.colab import files
uploaded = files.upload()
sha256_file = [f for f in uploaded if f.endswith('.sha256')][0]
private_key_file = [f for f in uploaded if f.endswith('.pem')][0]
print(f'Using hash file: {sha256_file}')
print(f'Using private key: {private_key_file}')
````

````python
# 🖋️ Sign the SHA-256 file
output_sig = sha256_file + '.sig'
!openssl dgst -sha256 -sign {private_key_file} -out {output_sig} {sha256_file}
print(f'Signature created: {output_sig}')
````

````python
# 📥 Download the signature file
files.download(output_sig)
````

---
### ✅ Verifying This Signature Later
Use this command with your public key on any Linux/macOS system:
```bash
openssl dgst -sha256 -verify public_key.pem -signature signed_document.sig signed_document.sha256
```
